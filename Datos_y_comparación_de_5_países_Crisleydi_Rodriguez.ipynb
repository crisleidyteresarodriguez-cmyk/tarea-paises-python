#***Datos y comparación de 5 países***




En este paso creé cuatro listas con los nombres de cinco países, sus poblaciones, sus PIB y sus áreas. Coloqué los datos en el mismo orden para que cada país tuviera su información correspondiente.

```python
# PASO 1: Crear una lista con los nombres de 5 países
paises = ["Estados Unidos", "China", "India", "Rusia", "Canadá"]

# Crear una lista con la población de cada país
# Los datos corresponden aproximadamente a 2024
poblaciones = [340000000, 1409000000, 1450000000, 144000000, 41000000]

# Crear una lista con el PIB de cada país en dólares
pib = [29180000000000, 18740000000000, 3910000000000,
       2170000000000, 2240000000000]

# Crear una lista con el área de cada país en kilómetros cuadrados
areas = [9833520, 9596961, 3287263, 17098246, 9984670]
```

Aquí organicé los países desde el que tiene mayor población hasta el que tiene menor población. Para hacerlo utilicé funciones de Python que me permitieron ordenar los datos y mostrarlos en pantalla.

```python
# PASO 2: Ordenar los países por población de mayor a menor
# Se agrupan los datos de cada país para mantenerlos relacionados
datos_paises = list(zip(paises, poblaciones, pib, areas))

# sorted() ordena los datos según la población (posición 1)
# reverse=True indica que el orden será descendente
datos_ordenados = sorted(datos_paises,
                         key=lambda pais: pais[1],
                         reverse=True)

# Mostrar los países ordenados por población
print("PAÍSES ORDENADOS POR POBLACIÓN:")
for pais in datos_ordenados:
    print(pais[0], "-", format(pais[1], ","), "habitantes")
```

En este paso dividí el PIB de cada país entre su población para saber cuánto representa por habitante. Después comparé los resultados para encontrar el país con el PIB per cápita más alto.

```python
# PASO 3: Encontrar el país con el PIB per cápita más alto
# El PIB per cápita se calcula dividiendo el PIB entre la población
pib_per_capita = []

for i in range(len(paises)):
    resultado = pib[i] / poblaciones[i]
    pib_per_capita.append(resultado)

# Encontrar la posición del valor más alto
posicion_mayor = pib_per_capita.index(max(pib_per_capita))

# Mostrar el país con el PIB per cápita más alto
print("\nPAÍS CON MAYOR PIB PER CÁPITA:")
print(paises[posicion_mayor])
print("PIB per cápita: $",
      format(pib_per_capita[posicion_mayor], ",.2f"))
```

Aquí busqué cuál de los cinco países tiene la mayor área y guardé sus datos en una tupla. Esta es diferente de una lista porque sus elementos no se pueden modificar después de crearla.

```python
# PASO 4: Crear una tupla inmutable con los datos
# del país que tiene la mayor área
posicion_area = areas.index(max(areas))

pais_mayor_area = (
    paises[posicion_area],
    poblaciones[posicion_area],
    pib[posicion_area],
    areas[posicion_area]
)

# Mostrar la tupla con los datos del país más extenso
print("\nPAÍS CON MAYOR ÁREA:")
print(pais_mayor_area)
```

Por último, mostré en pantalla la información de los cinco países, incluyendo su población, PIB y área. También organicé la forma en que aparecen los números para que sean más fáciles de leer.

```python
# PASO 5: Mostrar un resumen de todos los datos
# Este resumen permite comprobar la información utilizada
print("\nRESUMEN DE LOS DATOS ORIGINALES:")

for i in range(len(paises)):
    print(
        paises[i],
        "| Población:", format(poblaciones[i], ","),
        "| PIB: $", format(pib[i], ","),
        "| Área:", format(areas[i], ","), "km²"
    )
```
